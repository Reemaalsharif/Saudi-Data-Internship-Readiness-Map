# Skills and Tools Analysis

## Objective

Analyze the most frequently requested skills and tools in Saudi data internship and job postings.

## Dataset Overview

This notebook uses the cleaned Saudi data jobs dataset to analyze the most requested skills and tools in data-related job and internship postings.

## Analysis Areas

This notebook focuses on:

- Required tools.
- Required technical skills.
- Market demand insights.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
url = "https://raw.githubusercontent.com/Reemaalsharif/Saudi-Data-Internship-Readiness-Map/main/data/saudi_data_jobs_cleaned.csv"

df = pd.read_csv(url)

df.shape

df.columns
```

```python
df.shape
```

```python
df.columns
```

```python
df[['skills', 'tools']].head()
```

## Tools Analysis

This section analyzes the most frequently mentioned tools in Saudi data-related job postings.

```python
tools_series = df['tools'].str.split(',')
tools_exploded = tools_series.explode()
tools_cleaned = tools_exploded.str.strip()
tool_counts = tools_cleaned.value_counts()

tool_counts.head(10)
```

## Top Tools Visualization

Visualizing the most requested tools to identify the technologies commonly required in Saudi data-related roles.

```python
top_10_tools = tool_counts.head(10)

top_10_tools.plot(kind='bar', figsize=(10, 5))

plt.title('Top 10 Tools in Saudi Data Jobs')
plt.xlabel('Tool')
plt.ylabel('Number of Mentions')
plt.xticks(rotation=45)
plt.show()
```

### Initial Insight: Tools

SQL is the most frequently mentioned tool in the dataset, followed by Python, Excel, and Power BI.

This highlights the importance of SQL, Python, and visualization tools as core foundations for students preparing for data-related careers.

## Skills Analysis

This section analyzes the most frequently mentioned skills in Saudi data-related job postings.

```python
skills_series = df['skills'].str.split(',')
skills_exploded = skills_series.explode()
skills_cleaned = skills_exploded.str.strip()
skill_counts = skills_cleaned.value_counts()

skill_counts.head(10)
```

## Top Skills Visualization

Visualizing the most requested skills to understand the main capabilities required in Saudi data roles.

```python
top_10_skills = skill_counts.head(10)

top_10_skills.plot(kind='bar', figsize=(10, 5))

plt.title('Top 10 Skills in Saudi Data Jobs')
plt.xlabel('Skill')
plt.ylabel('Number of Mentions')
plt.xticks(rotation=45)
plt.show()
```

### Initial Insight: Skills

The most frequently mentioned skills include trend analysis, data validation, data analysis, data governance, reporting, and problem solving.

This shows that Saudi data-related roles require a combination of technical analysis abilities and business-focused skills such as reporting and problem solving.

## Summary

Key observations:
